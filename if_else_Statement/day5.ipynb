```python
sal = int(input("Enter Basic Salary: "))

if sal < 20000:
    hr = sal * 20 / 100
    da = sal * 10 / 100

elif sal <= 50000:
    hr = sal * 25 / 100
    da = sal * 15 / 100

else:
    hr = sal * 30 / 100
    da = sal * 20 / 100

total = sal + hr + da

if total > 100000:
    tax = total * 10 / 100
else:
    tax = total * 5 / 100

net = total - tax

print(f"Your basic salary: {sal}")
print(f"Your house rent: {hr}")
print(f"Your DA: {da}")
print(f"Your gross salary: {total}")
print(f"Your tax: {tax}")
print(f"Your net salary: {net}")
```
