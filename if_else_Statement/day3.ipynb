```python
#Volume of a Sphere
#V = 4/3 * π * r³
redius = float(input("Enter Redius"))
pi = 3.14
sphere = (pi*redius**3)*4/3
print(sphere)
```

```python
#Volume of a Cone
#V = 1/3 * π * r² * h
r = float(input("Enter redius"))
h = float(input("Enter Height"))
pi = 3.14
cone = 1/3*pi*r**2*h
print(cone)
```

```python
#Heron's Formula
a = int(input("Enter value of a"))
b = int(input("Enter value of b"))
c = int(input("Enter value of c"))
s = (a + b + c) / 2
print(s)
```

```python
#Compound Interest
#A = P * (1 + R/100) ** T
#CI = A - P
p = float(input("Enter principal Amount"))
r = float(input("Enter rate of percentage"))
t = float(input("Enter Time in years"))
a = p * (1 + r/100) ** t
print(f" Your Final Amount is: {a}")
ci = a-p
print(f"Your Componunt Intrest is: {ci}")
```
