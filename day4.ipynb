```python
#Leap year
#year % 400 == 0
year  = int(input("Enter Year"))
if (year % 400 == 0  or (year % 4 == 0 and year % 100 != 0)):
    print(f"{year}: This is Leap Year")
else:
    print(f"{year}: This is Not leap Year")
```

```python
#Character vowel or consonant
#a, e, i, o, u → Vowel
#Otherwise → Consonant
alp = str(input("Enter Any Alphabet A-Z : "))
vowel = ("a","e","i","o","u")
if(alp in vowel):   # list or group ke liye (in ) use karte hai
    print("This is Vowel")
else:
    print("This is Consonant")
```
