```python
# 2. Find the second-highest marks
marks = {
    "Amit": 75,
    "Rahul": 92,
    "Priya": 85,
    "Neha": 88
}
high = 0
second_high = 0
for key, values in marks.items():
    if values > high :
        second_high = high
        high = values
    elif second_high < values and second_high!=high:
        second_high = values
        
print(f"SECOND HIGHEST MARKS: {key,second_high}")
```

```python
# 3. Merge two dictionaries and add common values
d1 = {"apple": 10, "banana": 20, "mango": 30}
d2 = {"banana": 5, "mango": 15, "orange": 25}
marge = {**d1,**d2} # marge two dictionarys key nad values
print(marge)

marged = dict(zip(d1,d2)) #   marge dictionary keys 
print(marged)
```

```python
# 4. Remove Duplicate Data in Dictionary
data = {
    "a": 10,
    "b": 20,
    "c": 10,
    "d": 30,
    "e": 20
}
result = {}

for key, value in data.items():
    if value not in result.values():
        result[key] = value
        
print(result)
```

```python
# 7. Find common items in two dictionaries
d1 = {"pen": 10, "book": 50, "bag": 500}
d2 = {"book": 50, "bag": 600, "pencil": 30, "pen":10}
common = {}
for key, values in d1:
    if values in d2:
        common[values] = d1[values]
print(common)
```

```python
# 10. Group words by their first letter
words = ["apple", "ant", "banana", "ball", "cat", "car"]
result={}
for word in words:
    first = word[0]
    if first in result:
        result[first].append(word)
    else:
        result[first] = [word]
print(result)
```
