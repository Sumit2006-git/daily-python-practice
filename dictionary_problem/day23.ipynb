```python
# 1. Find the student with the highest marks
marks = {"Amit": 75, "Rahul": 92, "Priya": 85}
large = 0 
topper = ""
for key, values in marks.items():
    if values>large:
        large = values
        topper = key
print(f"Highest Marks: {topper,large}")
        
```

```python
# 2. Find the student with the lowest marks
marks = {"Amit": 75, "Rahul": 92, "Priya": 85}
low = float("inf") 
avg = ""
for key, values in marks.items():
    if low > values : 
        low = values
        avg = key
print(f"Lowest Marks:{avg,low}")
```

```python
# 4. Count students scoring above average
marks = {"Amit": 70, "Rahul": 90, "Priya": 80, "Neha": 60}
total = 0
for key, values in marks.items():
    total += values
    avg = total/len(marks)
count = 0
for key, values in marks.items():
    if avg < values:
        count += 1
       
print(f"Student Scoring Above The Average Marks: {count}")
    
    
```

```python
# 5. Sort dictionary by values
marks = {"Amit": 70, "Rahul": 90, "Priya": 80, "Neha": 60}
sorted_marks = sorted(
    marks.items(),
    key = lambda item : item[1],
    reverse = True
    
)
print(sorted_marks)
    
```

```python
# 6. Remove students who failed
marks = {"Amit": 75, "Rahul": 30, "Priya": 85, "Neha": 25}
pass_stu = {}
for key, values in marks.items():
    if values >= 40 :
        pass_stu[key] = values
print(pass_stu)     
```

```python
# 7. Count frequency of each number
numbers = [2, 3, 2, 5, 3, 2, 5, 7]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(f"Frequency of Numbers: {freq}")
```

```python
# 8. Find common keys in two dictionaries
d1 = {"name": "Amit", "age": 20, "city": "Indore"}
d2 = {"age": 21, "city": "Bhopal", "course": "Python"}
common = {}
for key in d1:
    if key in d2:
        common[key] = d1[key]
print(common)
            
```

```python
# 9. Calculate total price of a shopping cart
cart = {"pen": (10, 3), "book": (50, 2), "bag": (500, 1)}
total = 0
for item, values in cart.items():
    price = values[0]
    quantity = values[1]
    item_price = price * quantity
    total = total + item_price

for item, (price, quantity) in cart.items():
    item_total = price * quantity
    print(item, "=", item_total)

print(f"TOTAL PRICE: {total}")
```
