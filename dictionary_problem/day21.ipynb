```python
# 5. Find the length of a dictionary
marks = {
    "Math": 85,
    "English": 78,
    "Science": 92,
    "Hindi": 80
}
print(f"Length of Dictionary: {len(marks)}")
```

```python
# 6. Check whether a key exists
person = {"name": "Rohit", "age": 20, "city": "Delhi"}
if "age" in person:
    print("Age = Key present in Dictionary")
else:
    print("Age = Key not present in Dictionary")
```

```python
# 7. Print all keys and values
marks = {"Math": 85, "English": 78, "Science": 92}
for key, values in marks.items():
    print(key, values)
```

```python
# 8. Find the total of all values
marks = {
    "Amit": 70,
    "Rahul": 80,
    "Priya": 90
}
total = 0
for values in marks.values():
    total = total + values
print(f"TOTAL MARKS= {total}")
```

```python
# 9. Count the frequency of characters
text = "hello"
freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1


print(freq)
    
```

```python
# 10. Create a dictionary from two lists
keys = ["name", "age", "city"]
values = ["Aman", 18, "Bhopal"]
student = dict(zip(keys,values))
print(student)
```
