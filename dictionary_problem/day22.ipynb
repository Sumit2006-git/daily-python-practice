```python
# 1. Delete a key from dictionary
student = {
    "name": "Amit",
    "age": 19,
    "city": "Indore"
}
student.pop("age")
print(student)
```

```python
# 2. Find the average of dictionary values
marks = {
    "Amit": 70,
    "Rahul": 80,
    "Priya": 90,
    "Neha": 60
}
total = 0
for values in marks.values():
    total  = total + values
    sum  = total / len(marks)

print(sum)
```

```python
# 3. Count passing and failing students
marks = {
    "Amit": 75,
    "Rahul": 32,
    "Priya": 45,
    "Neha": 28,
    "Ravi": 60
}
pass_count = 0
fail_count = 0
for name,marks in marks.items():
    if marks >= 40: 
        pass_count = pass_count + 1
    else:
        fail_count = fail_count + 1
print(f"PASS STUDENTS {pass_count}") 
print(f"FAIL STUDENTS {fail_count}")
```

```python
# 5. Merge two dictionaries
dict1 = {"a": 10, "b": 20}
dict2 = {"c": 30, "d": 40}
marge = {**dict1, **dict2}
print(marge)
```

```python
# 6. Swap keys and values
data = {
    "A": 1,
    "B": 2,
    "C": 3
}
new = {}
for alpha, num in data.items():
    new[num] = alpha
print(new)
```

```python
# 7. Find duplicate values
marks = {
    "Amit": 80,
    "Rahul": 90,
    "Priya": 80,
    "Neha": 70,
    "Ravi": 90
}
duplicate = {}
for name, mark in marks.items():
    if mark in duplicate:
        duplicate[mark].append(name)
    else:
        duplicate[mark] = [name]
for mark, names in duplicate.items():
    if len(names) > 1:
        print(f"{mark}: {', '.join(names)}")
```

```python
# 8. Count words in a sentence
text = "python is easy and python is powerful"
empty = {}
for word in text.split():
    if word in empty:
        empty[word] += 1
    else:
        empty[word] = 1
print(empty)

text = "apple banana apple mango banana apple"
khali = {}
for words in text.split():
    if words in khali:
        khali[words] += 1
    else:
        khali[words] = 1
print(khali)
```

```python
# 9. Create a dictionary of squares
sqr = {}
for n in range(1,6):
    sqr[n] = n*n
print(sqr)   
```

```python
# 10. Group students by their result
marks = {
    "Amit": 75,
    "Rahul": 32,
    "Priya": 45,
    "Neha": 28,
    "Ravi": 60
}
result = {
    "pass" : [],
    "fail" : []
}
for name, values in marks.items():
    if values >= 40:
        result["pass"].append(name)
    else:
        result["fail"].append(name)
print(result)
```
