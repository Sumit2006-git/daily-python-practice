```python
# 1. Find the student with the highest average
students = {
    "Amit": {"math": 80, "science": 70, "english": 70},
    "Rahul": {"math": 95, "science": 85, "english": 90},
    "Priya": {"math": 75, "science": 80, "english": 85}
}
highest_avg = 0
topper = ""
for name, subjects in students.items():
    total = sum(subjects.values())
    average = total / len(subjects)

    if average > highest_avg:
        highest_avg = average
        topper = name

print(f"HIGHEST AVERAGE: {highest_avg}")
print(f"TOPPER STUDENT: {topper}")
    
```

```python
# 2. Group students by pass/fail result
marks = {
    "Amit": 75,
    "Rahul": 30,
    "Priya": 85,
    "Neha": 25,
    "Ravi": 45
}
result = {
    "pass" : [],
    "fail" : []
}
for name , numbers in marks.items():
    if  numbers < 40:
        result["pass"].append(name)
    else:
        result["fail"].append(name)

print(result)
```

```python
# 3. find duplicate values in dictionary
data = {
    "a": 10,
    "b": 20,
    "c": 10,
    "d": 30,
    "e": 20,
    "f": 10
}
result = {}
for key, value in data.items():
    if value not in result:
        result[value] = []

    result[value].append(key)
    
for value, keys in result.items():
    if len(keys) > 1:
        print(value, ":", keys)
```

```python
# 4. Sort a dictionary by value without sorted() or lambda
marks = {"Amit": 75, "Rahul": 92, "Priya": 60, "Neha": 85}

items = list(marks.items())

for i in range(len(items)):
    for j in range(i + 1, len(items)):
        if items[i][1] > items[j][1]:
            items[i], items[j] = items[j], items[i]

result = dict(items)
print(result)
```

```python
# 5. Find the most frequent word
text = "python is easy and python is powerful and python is popular"
freq = {}
for word in text.split():
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1
print(freq)
```
