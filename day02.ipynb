```python

sub1 =float(input("Enter mask on 1. Hindi"))
sub2 =float(input("Enter mask on sub 2. English"))
sub3 =float(input("Enter mask on sub 3. Mathas"))
sub4 =float(input("Enter mask on sub 4. physics"))
sub5 =float(input("Enter mask on sub 5. Chemistry"))
total= sub1+sub2+sub3+sub4+sub5
per= total*100/500
if(per>=75):
    print("Your Garde is  +A")
elif(per>=60):
    print("Your Grade is  A")
elif(per>=50):
    print("Your Grade is  B")
else:
    print("your Grade is  C")
print(f"Your percentage is: {per}%")
```

```python
# BMI CALCULATOR
w = float(input("Enter Your Body Weight"))
h = float(input("Enter Your Height in Meter"))
bmi = w/h**2
print(f"BMI = {bmi}")
if (bmi<=18.5):
    print("Under Weight")
elif (bmi<=24.9):
    print("Normal Weight")
elif (bmi<=29.9):
    print("Over Weight")
else:
    print("Obesity")
```
