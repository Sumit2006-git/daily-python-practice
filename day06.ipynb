```python
# Power of numbers M**n
m = int(input("Enter value of m"))
n = int(input("Enter value of n"))
i = 1
res = 1
while(i<=n):
    res=res*m
    i=i+1
print(res)
```

```python
# Fectorial Number
n = int(input("Enter Any Number: "))
fact = 1
i = 1
while i <= n:
    fact = fact * i
    i = i + 1
print(fact)
```

```python
#write program to print fibonicy series
n1 = 0
n2 = 1
for f in range(0,10):
    print(n1)
    f = n1+n2
    n1 = n2
    n2 = f
print(f)
```

```python
# write a progarm to check user given number prime or not
n = int(input("Enter Any Number"))
for i in range(2,n-1):
    if n%i == 0:
        print("Number is Not Prime")
        break
else:
    print("Number is Prime")
    
```
