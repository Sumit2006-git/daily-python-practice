```python
#List numbers = [2, 4, 6, 8, 10] find square print of all numbers
numbers = [2, 4, 6, 8, 10]
for i in numbers:
    i = i * i
    print(i)
    
```

```python
#List numbers = [10, 20, 30, 40, 50] find sum of all numbers
numbers = [10, 20, 30, 40, 50]
total = 0
for i in numbers:
    total = total+i
print(total)
```

```python
#List numbers = [11, 24, 35, 42, 53, 60] even numbers print in list.
numbers = [11, 24, 35, 42, 53, 60]
for i in numbers:
    if i%2==0:
        print("Even= ",i)
    
```

```python
#List numbers = [10, 25, 30, 45, 50, 65] odd numbers print in list.
numbers = [10, 25, 30, 45, 50, 65] 
for i in numbers:
    if i%2==0:
        print("odd= ",i)
```

```python
#List numbers = [10, 5, 20, 8, 30, 15] find largest value in this list without use of Max()function
numbers = [10, 5, 20, 8, 30, 15]
largest = 0
for i in numbers:
    if i>largest:
        largest = i
print(f"Largest Number: {largest}")
```

```python
#List numbers = [10, 5, 20, 8, 30, 15] find smallest value in this list
numbers = [10, 5, 2, 20, 8, 30, 15]
small = len(numbers)
for i in numbers:
    if i<small:
        small = i
print(f"Smallest Number is: {small}")
```

```python
#List numbers = [2, 5, 7, 10, 12, 15] me kitne even numbers hain, count karo.
numbers = [2, 5, 7, 10, 12, 15]
count = 0
for i in numbers:
    if i%2 == 0:
        count += 1
print(count)
```

```python
#List numbers = [10, 20, 10, 30, 10, 40] me 10 kitni baar present hai, count karo.
numbers = [10, 20, 10, 30, 10, 40, 10, 10]
count = 0
for i in numbers:
    if i==10:
        count += 1
print(count)
```
