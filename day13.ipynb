```python
#1. Find Common Element in Two List 
def common():
    list1 = [10, 20, 30, 40]
    list2 = [20, 40, 50, 60]
    common = []
    for i in list1:
        for j in list2:
            if i == j :
                common.append(i)
    print(common)

common()
```

```python
#2. Write a Python program to find the HCF (Highest Common Factor) of two numbers using a function.
HCF = 1
def hcf(a,b):
    i = 1
    while i<=a:
        if a % i == 0 and b % i == 0:
            HCF = i
        i += 1
    return f"Highest Common Factor = {HCF}"

print(hcf(24,60))
    
    
```

```python
# 3. find Duplicate Number 
numbers = [10, 20, 30, 20, 40, 10, 50]
def duplicate(numbers):
    for i in range(len(numbers)):
        for j in range(i+1,len(numbers)):
            if numbers[i]==numbers[j]:
                
                print(numbers[i])        
duplicate(numbers)
```

```python
# 4. Find Common Element in Two lists
l1 = [1, 2, 3, 4, 5]
l2 = [3, 2, 6 ,5 ,8]
def common_element(a,b):
    common = []
    for i in a:
        for j in b:
            if i == j:
                common.append(i)
    return common

c = common_element(l1,l2)
print(f"Common Elements Is: {c}")
```

```python
# 5. Write a Python program to swap the first and last elements of a list using their indexes and a function.
numbers = [10, 20, 30, 40, 50]
def swap():
    first = numbers[0]
    last_index = len(numbers) - 1
    last = numbers[last_index]

    numbers[0] = last
    numbers[last_index] = first
    
    print(numbers)

swap()
```

```python
# 7. Reverse List Using Index
num = [10, 20, 30, 40, 50]
def index(numbers):
    
    i = 0
    j = len(num) - 1
    while i < j:
        temp = num[i]
        numbers[i] = num[j]
        num[j] = temp
    
        i = i + 1
        j = j - 1
        return num
           
index(num)
        
```

```python
# 8. Seat Allocation System
seats = [1, 2, 3, 4, 5, 6, 7, 8]
seat = int(input("Enter Seat Number: "))

if seat in seats:
    print("Seat Available")
    seats.remove(seat)
    print("Seat Booked Successfully")
else:
    print("Seat Not Available")
print("Available Seats:", seats)
```

```python
# 9.Login System locked  🔐
def login():
    correct_pin = 2214
    attempt = 0
    while attempt < 3:
        pin = int(input("Enter Pin: "))
        if pin == correct_pin:
            print("CORRECT PIN")
            break
        else:
            attempt = attempt + 1
            print("Wrongh PIN !")

    if  attempt == 3:
        print("LOCKED !")

login()
```

```python
# 10. Write a function that counts how many times each number occurs
def frequecy(n):
    visited = []
    for i in n:
        if i not in visited:
            count = 0
            for j in n :
                if i == j:
                    count += 1

            print(i, "occurs", count,"times")
            visited.append(i)

f= frequecy([2, 3, 2, 5, 3, 2])
print(f)
```
