```python
# print prime number 1 to 100 in a list
prime_number = []
for i in range(2, 101):
    count = 0
    for num in range(1, i + 1):
        if i % num == 0:
            count += 1
    if count == 2:
        prime_number.append(i)

print(prime_number)
```

```python
# write of program to creat list of reindom number to sepret even or odd number list
l = [23,35,46,1,2,4,67,44,45,80]
o =[]
e =[]
for i in l:
    if i%2==0:
        e.append(i)
    else:
        o.append(i)
print("Original list:",l)
print("Even Number list:",e)
print("OOD Number list:",o)
```

```python
# Find Second Largest Number
num = [10, 45, 23, 87, 67, 89, 34]
largest = num[0]
second = num[0]
for i in num:
    if i>largest:
        second = largest
        largest = i
    elif second<i and i!= largest:
        second = i
print(f"Largest number is: {largest}")
print(f"Second Largest Number is: {second}")
   
```

```python
# Find Second Smallest Number
num = [45, 12, 78, 23, 13, 56, 34]
small = num[0]
second = num[0]
for i in num:
    if i<small:
        second = small
        small = i
    elif second > i and i!= small:
        second = i
        
print(f"Second Smallest Number is: {second}")
print(f"Smallest Number is: {small}")
```

```python
# Print reverce Number 
n = int(input("ENTER ANY NUMBERS:"))
rev = 0
while n > 0:
    digit = n % 10
    rev = rev*10+digit
    n = n//10
print(rev)
```

```python
# Count Vowels in str
name = input("Enter string: ")
vowel = ['a','e','o','i','u']
count = 0
for i in name:
    for j in vowel:
        if i == j:
            count += 1
print(f"Vowel in this Word: {count}")
```
