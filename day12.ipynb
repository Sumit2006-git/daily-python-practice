```python
# 1. write a function find duplicate(numbers) that return the duplicate numbers
l = [23, 45, 67, 12, 87, 98, 23, 45, 98]
def find_duplicate(numbers):
    duplicate = []
    for i in numbers:
        if numbers.count(i) > 1 and i not in duplicate:
            duplicate.append(i)
    print(duplicate)

find_duplicate(l)
    
```

```python
# 2. Find Third Largest Number in a list
numbers = [23, 45, 65, 12, 87, 98, 56]
def third_large(numbers):
    large1 = 0
    second2 = 0
    third3 = 0
    for i in numbers:
        if i > large1:
            third3 =second2
            second2 = large1
            large1 =i
        elif i > second2:
            third3 = second2
            second2 = i
        elif i > third3:
            third3 = i

    print(f"Third Largest Number Is: {third3}")
third_large(numbers)
            
```

```python
# 3. Find Third Smallest Number in list
l = [67, 45, 65, 12, 87, 98, 56]
def third_small(l):
    first_small = l[0]
    second_small = l[0]
    third_small = l[0]
    for i in l:
        if i < first_small:
            third_small = second_small
            second_small = first_small
            first_small = i
        elif i < second_small:
            third_small = second_small
            second_small = i
        elif i < third_small:
            third_small = i
    return f"Third Smallest Number Is: {third_small}"

print(third_small(l))
            
```

```python
# 5.Separate Positive and Negative Numbers
l = [10, -5, 8, -2, 0, 7]
def separate_n(l):
    positive = []
    nagitive = []
    for i in l:
        if i > 0:
            positive.append(i)
        if i < 0:
            nagitive.append(i)

    print(f"Positive = {positive}")
    print(f"Nagitive = {nagitive}")
separate_n(l)
```

```python
# 6. find missing number in a list
numbers = [1, 2, 3, 5, 6, 7, 8]
def missing(numbers):
    n = len(numbers) + 1
    total = n*(n+1)//2
    sum_list = 0
    for i in numbers:
        sum_list = sum_list + i
    missing = total - sum_list
        
    # print(sum_list)
    # print(total)
    print(f"Missing Number is: {missing}")

missing(numbers)
```

```python
# 7. Search Number in a list
numbers = [12, 43, 56, 100, 55, 78, 92, 87, 34, 28]
num = int(input("Search Number: "))

def search(numbers):
    for i in numbers:
        if i == num:
            print(f"Searching Comlete Your Number Is: {num}")

search(numbers)
```

```python
#“Write a Python program to create a simple bank account system using a function. 
#The user can deposit money, withdraw money, and check the account balance.”
def bank_account():
    balance = 10000

    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")

    choice = int(input("Enter Your Choice: "))
    if choice == 1:
        amount = int(input("Enter Deposit Amount: "))
        balance = balance + amount
        print("Deposit Successful")
        print("Current Balance:", balance)
    elif choice == 2:
        amount = int(input("Enter Withdraw Amount: "))
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal Successful")
            print("Remaining Balance:", balance)
        else:
            print("Insufficient Balance")
    elif choice == 3:
        print("Current Balance:", balance)
    else:
        print("Invalid Choice")


bank_account()
```
