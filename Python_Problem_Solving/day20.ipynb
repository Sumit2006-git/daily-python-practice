```python
# 1. Count the frequency of  element in a list.
numbers = [4, 2, 4, 5, 2, 4, 7, 5, 2, 8]
def frequency(n):
    count = 0
    for i in numbers:
        if i==n:
            count+=1

    print(count)
n = int(input("Find Frequecy for any Number:"))
frequency(n)
    
```

```python
# 2. Count the frequency of every element in a list.
numbers = [4, 2, 4, 5, 2, 4, 7, 5, 2, 8]
def frequency(n):
    freq = {}
    for i in n:
        if i in freq:
            freq[i]+=1

        else:
            freq[i]=1

    return freq
frequency(numbers)
        
```

```python
# 3, Write a Python program to check whether a given number is a palindrome or not
def pelindrom(n):
    original = n
    rev = 0
    while n>0:
        digit = n % 10     # last digit
        rev = rev*10+digit  
        n = n // 10

    if original == rev:
        return True

    else:
        return False
n = int(input("Enter Any Number to check pelindrom NO."))
pelindrom(n)
```

```python
# 4. Write a Python program to reverse a given number using a function and a while loop.
def reverce(num):
    rev = 0
    while num > 0:
        digit = num % 10   # last digit calculate
        rev = rev*10 + digit   # sepret the last digit
        num = num // 10     # last digit are remove
    print(rev)
numbers = int(input("Enter Any number to Reverce"))
reverce(numbers)
        
```

```python
# 5.Write a Python program to check whether a given number is an Armstrong Number
def armstrong(n):
    original = n
    temp = n
    count = 0

    while temp > 0:
        count += 1 # count the number of digit
        temp = temp // 10 # last digit remove

    total = 0
    while n > 0 :
        digit = n % 10
        total = total + digit**count
        n = n // 10

    if original == total:
        return True
    else:
        return False
        
n = int(input("Enter Any Number to Check Armstrong NO."))
armstrong(n)
    
```

```python
# 6. Write a Python program to check whether a given number is a Perfect Number
def perfect(n):
    total = 0
    for i in range(1,n):
        if n % i :
            total = i + 1
    if total == n :
        return True
    else:
        return False

num = int(input ("Enter Any Number to Check Perfect NO."))
perfect(num)
```

```python
# 7. Write a Python program to check whether a given number is a Strong Number
def strong(number):
    original = number
    total = 0
    while number > 0:
        digit = number % 10
        fact = 1
        i = 1
        while i <= digit:
            fact = fact * i
            i += 1
        total = total + fact
        number = number // 10
    if original == total:
        return True
    else:
        return False

if strong(145):
    print("Strong Number")
else:
    print("Not a Strong Number")
```

```python
# 8. Write a Python program to print the Fibonacci series up to n terms
def fibonnacci(n):
    a = 0
    b = 1
    count = 0
    while count < n:
        print(a)
        c = a+b
        a = b 
        b = c
        count += 1

fibonnacci(10)
        
```

```python
# 9. Write a Python program to check whether a given number is a Prime Number
def prime(n):
    count = 0
    for i in range(1,n + 1):
        if n % i == 0:
            count += 1
    if count == 2:
        print("Numbers is Prime")
    else:
        print("number is not prime")

prime(8)
        
```

```python
# 10. Write a Python program to check whether a given number is an Automorphic Number
# (5^2 = 25), aur 25 ka last digit 5 hai.
def auto(num):
    original = num
    sqr = num * num
    count = 0
    while num > 0:
        count += 1
        num = num // 10
        last_part = sqr % (10 ** count)
    if original == last_part:
        print("Number is Automorphic")
    else:
        print("Number is NOT Automoephic")
num = int(input("Enter Any Number to Check Automorphic Number:"))
auto(num)
```
