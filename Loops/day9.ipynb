```python
# Sum of list
l = [23, 22, 11, 67, 56, 10, 11]
total = 0
for i in l :
    total = total + i
print(f" Sum of All Element Is : {total}")
```

```python
# Sum of Even Number in a List
l = [1, 2, 3, 4, 5, 6, 7, 8, 9, 6]
total  = 0
for i in l:
    if i % 2 == 0 :
        total = total + i

print(f" Sum of Even Numbers: {total}")
```

```python
# Sum of OOD Number in a List
l = [1, 2, 3, 4, 5, 6, 7, 8, 9, 6]
total  = 0
for i in l:
    if i % 2!= 0 :
        total = total + i

print(f"Sum of OOD Numbers: {total}")
```
