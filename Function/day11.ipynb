```python
# 6. Find Armstrong Number
def arm(n):
    original = n
    temp = original
    arm_n = 0
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    while temp > 0:
        digit = temp % 10
        arm_n = arm_n + digit**count
        temp = temp // 10
    if arm_n == original:
        print(f"THIS NUMBER IS ARMSTRONGH NUMBER: {original}")
    else:
        print(f"THIS NUMBER IS NOT ARMSTRONGH NUMBER: {original}")

arm(153)
```

```python
# 7. ATM Withdrawal
def atm(amount,balance):
    if amount > balance:
        print("Insufficient Balance")
    elif amount % 100!= 0:
        print("Amount is Not Valid")
    elif amount > 20000:
        print("Withdrawal Not Valid")
    else:
        balance = balance - amount
        print("Withdrawal Successful")
        print(f"Availabe Balance: {balance}")
        
amount = int(input("Enter Withdrawal Amount:"))
atm(amount,50000)       
    
```

```python
# 8. Numbers Frequency Counter
def frequency(n,digit):
    count = 0
    while n > 0:
        last_d = n % 10
        if digit == last_d:
            count += 1
        n = n // 10
    print(f"Frequency: {count}")
n = int(input("Enter frequency numbers:"))
digit = int(input("Enter Digit"))
frequency(n,digit)
```
