```python
# 1 . Write a lambda function that takes two numbers and returns their sum.
add = lambda a,b:a+b
result = add(40,40)
print(result)
```

```python
# 2 . Write a lambda function that returns the square of a number.
sqr = lambda n : n**2
result = sqr(20)
print(result)
```

```python
# 3 . Write a lambda function that returns the cube of a number.
cube = lambda c : c**3
result = cube(10)
print(result)
```

```python
# 4 . Write a lambda function that returns "Even" if a number is even, otherwise "Odd".
even = lambda n :"even" if n % 2 == 0  else "odd"
res = even(155)
print(res)
```

```python
# 5 . Write a lambda function that takes two numbers and returns the larger number.
great = lambda a,b : "a is greter then b" if a>b  else "b is greter then a"
output = great(12,15)
print(output)
```

```python
# 6 . Write a lambda function that takes two numbers and returns the smaller number.
small = lambda a,b : "a is smaller then b" if a<b  else "b is smaller then a"
output = small(12,15)
print(output)
```

```python
# 7 . Write a lambda function to convert Celsius into Fahrenheit.
fehrenheit = lambda c : c*9/5+32
result = fehrenheit(45)
print(result)
```

```python
# 8 . Write a lambda function that returns the last digit of a number.
last_digit = lambda n : n % 10
result = last_digit(12345)
print(result)
```

```python
# 9. Check positive, negative, or zero
check = lambda x: f"{x} is a positive number" if x > 0 else (f"{x} is a negative number" if x < 0 else f"{x} is zero")

print(check(1))
print(check(-3))
print(check(0))
```

```python
# 10. Find Maximum for a list by use a lambda function
numbers = [10, 45, 23, 89, 12]
maximum = max(numbers, key=lambda x:x)
print("Muximum:", maximum)
```
