```python
# 1 .Use reduce() + lambda to find the total number of characters in all the words.
from functools import reduce
words = ["Python", "AI", "Machine", "Learning"]
count = reduce(lambda x,y : x + len(y), words , 0)
print(count)
```

```python
# 2. Use reduce() + lambda to find the longest word.
words = ["Python", "Artificial", "AI", "Machine", "Learning"]
longest = reduce(lambda x,y : x if len(x) > len(y) else y , words)
print(longest)
```

```python
# 3. Use reduce() + lambda to find the longest word.
words = ["Python", "Artificial", "AI", "Machine", "Learning"]
sort = reduce(lambda x,y : x if len(x) < len(y) else y , words)
print(sort)
```

```python
# 4. Student Marks Processing System
# filter() — Select only marks greater than or equal to 40.
# map() — Add 5 bonus marks to each selected mark.
# reduce() — Calculate the total of all the updated marks.

from functools import reduce
marks = [25, 45, 60, 32, 75, 38, 90, 50]
greater = list(filter(lambda x : x >= 40 , marks))
bonus = list(map(lambda x : x+5 , greater))
total = reduce(lambda x,y : x+y , bonus )
print("Pass Student = ", greater)
print("Bonus Numbers =", bonus)
print("Total Marks =",total)
```

```python
# 5.Use filter(), map() and reduce() together in one Python program.
# filter() — Select salaries greater than or equal to ₹25,000.
# map() — Increase each selected salary by 10%.
# reduce() — Calculate the total of all updated salaries.

from functools import reduce
salary = [18000, 25000, 32000, 15000, 45000, 28000, 12000]
salary25000 = list(filter(lambda x : x >= 25000 , salary))
increase = list(map(lambda x : x + 10*x/100 , salary25000))
total = reduce(lambda x,y : x+y , increase)
print("Salary Greater then 25000=", salary)
print("Increase 10% Salary=",increase)
print("Total Salary=",total)
```
