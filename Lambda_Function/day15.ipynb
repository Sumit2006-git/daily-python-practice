```python
#1.Write a Python program to find the length of a given string using a lambda function.
length = lambda x:len(x)
print(length("sumit"))
```

```python
#2.lambda + sorted() ka use karke strings ko length ke according ascending order me sort karo
words = ["Python", "AI", "Machine", "ML"]
result = sorted(words, key=lambda x:len(x))
print(result)
```

```python
#3.lambda + filter() ka use karke list me se woh numbers nikalo jo 5 se completely divisible hain.
numbers = [10, 13, 25, 31, 40, 47, 55, 62]
five = filter(lambda x:x%5==0,numbers)
print(list(five))
```

```python
#4.filter() and lambda to extract all the even numbers from the given list.
numbers = [10, 13, 25, 31, 40, 47, 55, 62]
even = filter(lambda x:x%2==0,numbers)
print(list(even))
```

```python
#5.filter() and lambda to extract all the OOd numbers from the given list.
numbers = [10, 13, 25, 31, 40, 47, 55, 62]
odd = filter(lambda x:x%2!= 0,numbers)
print(list(odd))
```

```python
#6.Write filter() and lambda to filter out all marks greater than 60 from the given list.
marks = [45, 72, 55, 89, 61, 38, 95, 50]
num = filter(lambda x:x>=60,marks)
print(list(num))
```

```python
#7.Write filter() and lambda to filter all names that start with the letter "A"
names = ["Aman", "Rahul", "Ankit", "Sumit", "Ajay", "Ravi"]
s = filter(lambda x:x[0] == "S",names)
print(list(s))
```

```python
#8.Write filter() and lambda to extract all three-digit numbers from the given list.
numbers = [12, 105, 45, 999, 67, 321, 8, 1000]
digit = filter(lambda x:100 <= x <= 999,numbers)
print(list(digit))
```

```python
#9.filter() and lambda to extract all numbers that are divisible by both 3 and 5.
numbers = [15, 20, 30, 45, 50, 60, 72, 90, 100]
divide = filter(lambda x:x%5==0 or x%3==0,numbers)
print(list(divide))
```

```python
#10.Write a Python program using filter() and lambda to extract numbers between 20 and 50.
numbers = [12, 25, 38, 55, 42, 67, 30, 18]
between = filter(lambda x : 20<x<50, numbers)
print(list(between))
```
