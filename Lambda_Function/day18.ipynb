```python
# 1. Use Map()+ lambda function to convert marsk into  Grade
marks = [95, 82, 67, 54, 39, 25]
result = list(map(lambda n : "A" if n>=90 else
                             "B" if n>=75 else
                             "C" if n>=60 else
                             "D" if n>=40 else
                            "Fail" , marks))
print(result)
```

```python
# 2. Use Map()+ lambda to convert every nagitive numbers in positive numbers
numbers = [-5, 10, -20, 15, -8]
result = list(map(lambda x : abs(x),numbers ))
print(result)
```

```python
# 3. Use map() + lambda to return:
# "Positive" if number > 0
# "Negative" if number < 0
# "Zero" if number == 0
numbers = [10, -5, 20, -8, 0, 15]
result = list(map(lambda x : "Positive" if x > 0 else "Nagitive" if x < 0 else "Zero" , numbers))
print(result)
```

```python
# 4. Use map() + lambda to get the last digit of every number.
numbers = [123, 456, 789, 1024, 555]
digit = list(map(lambda x : x % 10, numbers))
print(digit)
```

```python
# 5. Use map() + lambda to convert every name into uppercase.
names = ["ujjawal", "rahul", "aman", "rohit"]
upper_case = list(map(lambda x : x.capitalize() , names))
print(upper_case)
```

```python
# 6. Use reduce() and lambda to calculate the sum of all numbers in the list.
from functools import reduce
num = [12, 23, 45, 67, 88 ,90]
result = reduce(lambda x,y : x+y , num)
print("SUM=",result)
```

```python
# 7. Use reduce() + lambda to find the largest number.
from functools import reduce
num = [12, 21, 10, 67, 11, 90, 5]
large  = reduce(lambda x,y : x if x > y else y , num)
print("Large=", large)
```

```python
# 9. Use reduce() + lambda to find the sum of only the even numbers.
from functools import reduce
numbers = [10, 15, 20, 25, 30, 35, 40]
even_sum = reduce(lambda x,y : x+y ,  filter(lambda x : x%2 == 0 , numbers))
print("Sum of Even No.",even_sum)
```

```python
# 10. Use reduce() + lambda to find the sum of only the even numbers.
from functools import reduce
numbers = [10, 15, 20, 25, 30, 35, 40]
odd_sum = reduce(lambda x,y : x+y , filter(lambda x: x%2!= 0 , numbers))
print("Sum of odd No.",odd_sum)
```
