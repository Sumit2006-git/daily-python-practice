```python
# 1. Use map() and lambda to find the square of each number.
num = [2, 4, 6, 8, 10]
result = list(map(lambda n : n**2 ,num))
print(result)

```

```python
# 2. Use map() and lambda to convert each temperature from Celsius to Fahrenheit.
celsius = [0, 10, 20, 30, 40]
f = list(map(lambda c : 9/5*c+32,celsius))
print(f)
```

```python
# 3.Use map() and lambda to add 5 bonus marks to every student's marks.
marks = [45, 62, 78, 89, 34, 95]
bonus = list(map(lambda m : m+5 , marks))
print(bonus)
```

```python
# 4. Use map() and lambda to increase every employee's salary by 10%.
salary = [25000, 32000, 45000, 50000, 28000]
employee = list(map(lambda salary : (salary*10/100) + salary , salary))
print(employee)
```

```python
# 5. Use map() and lambda to calculate the final price after a 20% discount on each product.
prices = [500, 1200, 750, 2000, 350]
discount = list(map(lambda price : price - (price*30/100),prices))
print(discount)
```

```python
# 6. Use map() and lambda to convert every name in the list to uppercase.
names = ["sumit", "rahul", "amit", "priya", "rohit"]
uppercase =  list(map(lambda n : n.upper(),names))
print(uppercase)
```

```python
# 7. find the length of each words use map() to find the length each word
words =  ["python" , "statistics", " Data" , " Ai"]
length =  list(map(lambda x : len(x), words))
print(length)
```

```python
# 8. use by map () function to convert feet into centimeters
feet  =  [1,2,3,4,6,7,8,9]
cm = list(map(lambda feet : feet*30.48 , feet))
print(cm)
```

```python
# 9. use map() function check even odd numbers
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_odd = list(map(lambda n : "Even" if n % 2 ==0 else "odd", numbers))
print(even_odd)
```

```python
# 10.  Use map() and lambda to check nagitive positive and zero numbers in the list 
numbers = [10, -56, 12, -1, 34, -12, 0, 0]
positive = list(map(lambda x : x>0, numbers ))
nagitive = list(map(lambda x : x<0, numbers))
zero = list(map(lambda x : x==0, numbers))
print("Positive= ",positive)
print("Nagitive= ",nagitive)
print("Zero= ", zero)
```
