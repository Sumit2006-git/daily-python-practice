```python
# 1. filter() and lambda to extract numbers whose last digit is 5.
num = [15, 22, 35, 41, 55, 67, 75, 80]
end = filter(lambda x:x % 10 == 5,num)  # x % 10 = last digit
print(list(end))
```

```python
# 2. filter() and lambda to extract numbers whose first digit is 1.
num = [15, 22, 35, 41, 55, 67, 75, 80]
start = filter(lambda x: x // 10 == 1 ,num)   # x // 10 = remove last digit
print(list(start)) 
```

```python
# 3. filter() and lambda to extract only two-digit numbers.
num = [1, 20, 4, 45, 6, 67, 9, 70, 8, 90, 0, 99]
digit = filter(lambda x: 10<x<100,num)
print(list(digit))
```

```python
# 4. filter() and lambda to extract numbers that are both positive and nagitive.
numbers = [-10, 15, 22, -8, 35, 40, -25, 18]
positive = filter(lambda n : n > 0 ,numbers)
nagitive = filter(lambda n : n < 0 ,numbers)
print("positive=",list(positive))
print("nagitive=", list(nagitive))
```

```python
# 5. filter() and lambda to extract numbers that are both three-digit and even.
num = [105, 120, 245, 301, 400, 512, 67, 999]
even = filter(lambda n : 100 < n < 999 and n % 2 == 0 , num)
print("Three Digit Even No = ", list(even))
```

```python
# 6. filter() and lambda to extract numbers that are divisible by 3 but not divisible by 5.
numbers = [15, 18, 20, 21, 30, 33, 40, 45, 52, 60]
divide = filter(lambda x : x % 3 == 0 and x % 5!= 0 , numbers)
print(list(divide))
```

```python
# 7. filter() and lambda to extract numbers whose last digit is greater than 5.
num = [23,23,366,29,55,12,5889]
digit = filter(lambda x : x % 10 > 5 , num)
print(list(digit))
```

```python
# 8. filter() and lambda to extract strings that contain more than 3 vowels.
words = ["education", "python", "beautiful", "computer", "india"]
result = filter(lambda x : sum(1 for ch in x if ch in "auieo")>3 ,words)
print("more then 3 vowels = ",  list(result))
```
