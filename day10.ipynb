```python
# 4. find the factorial of n (n is the peremeter) by using function
def fact(n):
    fact = 1
    for i in range(1, n + 1) :
        fact = fact * i
    print(fact)

fact(12)        
```

```python
# 6. Check even or odd numbers by Using Function
def even_odd(n):
    if n % 2 == 0:
        print(f"Number is even -> {n}")
    else:
        print(f"Number is odd -> {n}")
even_odd(12)
```

```python
# 7.  Prime Numbers (Difference) by Using Function
def prime(n):
    for i in range(2,n):
        if n % i == 0:
            return False
        else:
            return True
prime(4)           
```

```python
# Check Prime Numbers Using By Function
def prime(n):
    count = 0
    for i in range(1,n+1):
        if n % i == 0:
            count += 1
    if count == 2:
        print("Numer is Prime")
    else:
        print("Number is Not Prime")
prime(7)            
```
